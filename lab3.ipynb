5.6.1 Reading, Parsing, and Analyzing Text File Data


```python

import math

class Points:
    def __init__(self, x=0.0, y=0.0):
        self.x, self.y = x, y


class Polyline:
    def __init__(self, points=None):
        self.points = points if points is not None else []

    def getLength(self):
        length = 0.0

        for i in range(len(self.points) - 1):
            dx = self.points[i + 1].x - self.points[i].x
            dy = self.points[i + 1].y - self.points[i].y
            length += math.sqrt(dx**2 + dy**2)

        return length


def readPolylineFile(fileName):
    polylines = {}
    currentPolyLine = None

    with open(fileName, 'r') as f:
        for line in f:
            line = line.strip()

            if not line or line.lower() == 'polyline;':
                continue

            # Detect the start of a new polyline
            if ':' in line:
                label, coords = line.split(':', 1)
                currentPolyLine = label.strip()
                polylines.setdefault(currentPolyLine, [])
            else:
                coords = line

            if currentPolyLine is None or ',' not in coords:
                continue

            # Remove trailing semicolon and parse coordinates
            coords = coords.strip().rstrip(';')
            xCoord, yCoord = coords.split(',', 1)

            point = Points(float(xCoord), float(yCoord))
            polylines[currentPolyLine].append(point)

    return polylines


polylines = readPolylineFile('polyline.txt')

pointsForFirstPoly = polylines['1']
pointsForSecondPoly = polylines['2']

print("The first polyline number is:", len(pointsForFirstPoly))
print("The total number of points:", 
      len(pointsForFirstPoly) + len(pointsForSecondPoly))

polyline1 = Polyline(pointsForFirstPoly)
length1 = polyline1.getLength()
print("The length of the first polyline is:", length1)

polyline2 = Polyline(pointsForSecondPoly)
length2 = polyline2.getLength()
print("The length of the second polyline is:", length2)
```
